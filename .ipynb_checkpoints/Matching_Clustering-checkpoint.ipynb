```python
import pandas as pd
import numpy as np
import sys, os
sys.path.append(os.path.abspath("functions"))
from server_config import DATA_PATH


from therapist_clustering import cluster_therapists_gower_kmedoids, repair_therapist_age
```

```python
df_all = pd.read_csv(DATA_PATH + "/toni_preprocessed.csv")
```

```python
therapist_numeric = ['t0_age_t','t0_experience_t', 't0_bfi_t_extraversion', 't0_bfi_t_agreeableness', 't0_bfi_t_conscientiousness', 't0_bfi_t_neuroticism', 't0_bfi_t_openness'
                     ]
therapist_categorical = ['t0_appliedmethod_t_VT', 't0_appliedmethod_t_TP','t0_appliedmethod_t_AP','t0_method_t_CBT','t0_method_t_PDT','t0_method_t_ST','t0_method_t_AP'
                            't0_appliedmethod_t_ST','t0_appliedmethod_t_Other','t0_gender_t','t0_relationship_t','t0_sexualorientation_t','t0_children_t']
```

```python
res = cluster_therapists_gower_kmedoids(
    df_all,
    therapist_numeric=therapist_numeric,
    therapist_categorical=therapist_categorical,
    K_grid=tuple(range(4, 13)),
    min_dyads_per_cluster=25,
    method="famd_kmeans",
    famd_components=8,
)
```

```python


def cluster_quality_from_distance(D, labels, medoids=None):
    """
    D: (n x n) precomputed distance matrix
    labels: array-like of cluster ids (not required to be 0..K-1)
    medoids: iterable of medoid indices aligned to sorted(unique(labels)),
             or None to skip medoid dispersion.
    """
    D = np.asarray(D, float)
    z = np.asarray(labels)
    n = len(z)
    uniq = np.array(sorted(np.unique(z)))
    K = len(uniq)

    # -------- safe silhouette with precomputed distances (handles singletons) --------
    def _safe_silhouette(D, z):
        n = len(z); labs = np.array(sorted(np.unique(z)))
        # precompute index sets
        idx_by_c = {c: np.where(z == c)[0] for c in labs}
        s_each = np.empty(n, dtype=float)
        for i in range(n):
            ci = z[i]
            same = idx_by_c[ci]
            # a(i): mean intra-cluster distance
            if len(same) <= 1:
                a = 0.0
            else:
                a = float(np.mean(D[i, same[same != i]]))
            # b(i): minimum mean distance to other clusters
            b = np.inf
            for c in labs:
                if c == ci: continue
                other = idx_by_c[c]
                if len(other) == 0:  # shouldn't happen
                    continue
                b = min(b, float(np.mean(D[i, other])))
            if not np.isfinite(b):  # degenerate case (all in one cluster)
                s_each[i] = np.nan
            else:
                m = max(a, b)
                s_each[i] = 0.0 if m == 0.0 else (b - a) / m
        # overall as mean of finite values
        s_overall = float(np.nanmean(s_each)) if np.isfinite(s_each).any() else np.nan
        return s_overall, s_each

    sil_overall, sil_each = _safe_silhouette(D, z)
    sil_by_cluster = (
        pd.DataFrame({"cluster": z, "s": sil_each})
        .groupby("cluster", as_index=False)["s"]
        .agg(sil_mean=lambda s: float(np.nanmean(s)),
             sil_median=lambda s: float(np.nanmedian(s)),
             min=lambda s: float(np.nanmin(s)),
             max=lambda s: float(np.nanmax(s)),
             count="size")
        .sort_values("cluster")
        .reset_index(drop=True)
    )

    # -------- intra-compactness & Dunn --------
    intra_rows = []
    inter_min = np.inf
    idx_by_c = {c: np.where(z == c)[0] for c in uniq}

    for c in uniq:
        idx = idx_by_c[c]
        if len(idx) <= 1:
            intra_vals = np.array([0.0]) if len(idx) == 1 else np.array([])
        else:
            M = D[np.ix_(idx, idx)]
            intra_vals = M[np.triu_indices_from(M, k=1)]
        intra_rows.append({
            "cluster": int(c),
            "intra_mean": float(np.mean(intra_vals)) if intra_vals.size else 0.0,
            "intra_median": float(np.median(intra_vals)) if intra_vals.size else 0.0,
            "intra_90pct": float(np.quantile(intra_vals, 0.90)) if intra_vals.size else 0.0,
            "intra_max_diameter": float(np.max(intra_vals)) if intra_vals.size else 0.0,
            "size": int(len(idx)),
        })
    intra_df = pd.DataFrame(intra_rows)

    # min inter-cluster distance (any pair of points)
    for i, ci in enumerate(uniq):
        idx_i = idx_by_c[ci]
        for cj in uniq[i+1:]:
            idx_j = idx_by_c[cj]
            if len(idx_i) == 0 or len(idx_j) == 0:
                continue
            inter_block = D[np.ix_(idx_i, idx_j)]
            inter_min = min(inter_min, float(np.min(inter_block)))

    denom = float(intra_df["intra_max_diameter"].max()) if len(intra_df) else np.nan
    dunn = float(inter_min / denom) if (np.isfinite(inter_min) and denom > 0) else np.nan

    # -------- medoid-centered dispersion (robust to non-contiguous labels) --------
    medoid_df = pd.DataFrame()
    if medoids is not None:
        medoids = np.asarray(medoids).ravel().astype(int)
        rows = []
        # assume medoids are aligned to sorted(unique(labels))
        if medoids.size == len(uniq):
            label_medoid_pairs = list(zip(uniq, medoids))
        else:
            # fallback: treat medoid list as aligned to 0..K-1, map by position if possible
            label_medoid_pairs = list(zip(uniq, medoids[:len(uniq)]))
        for c, m_idx in label_medoid_pairs:
            members = idx_by_c.get(c, np.array([], dtype=int))
            if members.size == 0:
                continue  # skip truly empty (shouldn't happen)
            dist_to_medoid = D[m_idx, members]
            rows.append({
                "cluster": int(c),
                "medoid_index": int(m_idx),
                "medoid_mean_dist": float(np.mean(dist_to_medoid)),
                "medoid_90pct_dist": float(np.quantile(dist_to_medoid, 0.90)),
                "medoid_max_dist": float(np.max(dist_to_medoid)),
                "size": int(members.size),
            })
        medoid_df = pd.DataFrame(rows).sort_values("cluster").reset_index(drop=True)

    return {
        "silhouette_overall": sil_overall,
        "silhouette_by_cluster": sil_by_cluster,
        "intra_compactness": intra_df,
        "dunn_index": dunn,
        "medoid_dispersion": medoid_df,
        "K": int(K),
        "n": int(n),
    }
```

```python
# from your clustering run:
labels = res["therapists_table"]["therapist_group"].to_numpy()
D       = res["distance_matrix"]
medoids = res["kmedoids_model"].medoid_indices_
```

```python

q = cluster_quality_from_distance(D, labels, medoids=medoids)

q["silhouette_overall"]                  # ~0.2–0.5 is common for real data; >0.5 is strong
q["silhouette_by_cluster"]               # check for clusters with low/negative means
q["intra_compactness"]                   # smaller intra_mean / diameter is better
q["dunn_index"]                          # higher is better (>= ~0.3 often decent)
q["medoid_dispersion"]                   # sanity-check cluster spread around exemplars
```
